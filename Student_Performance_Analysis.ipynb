## Univariate Analysis

Univariate analysis is performed to understand the distribution of individual variables. The following plot shows the distribution of math scores.


```python
plt.figure(figsize=(6,4))
sns.histplot(df['math score'],kde=True)
plt.title('Distribution of Math Scores')
plt.xlabel('math scores')
plt.ylabel('Frequency')
plt.show()
```

### Insight

The math score distribution appears approximately normal, with most students scoring around the average range.


```python
plt.figure(figsize=(6,4))
plt.boxplot(df[['math score','reading score','writing score']])
plt.title("Score distribution comparison")
plt.xticks([1,2,3],['math','reading','writing'])
plt.ylabel('score')
plt.show()
```

## Bivariate Analysis: Gender and Math Performance

This section analyzes the relationship between gender and math scores to identify any performance differences.


```python
plt.figure(figsize=(6,4))
sns.boxplot(x='gender',y='math score',data=df)
plt.title("Math Score by Gender")
plt.show()
```

### Insight

Male students show a slightly higher median math score compared to female students, although the distributions largely overlap.


## Impact of Test Preparation on Performance

This analysis examines whether completing a test preparation course affects student math scores.


```python
plt.figure(figsize=(6,4))
sns.boxplot(x='test preparation course',y='math score',data=df)
plt.title('Math score by Test preparation')
plt.show()
```

### Insight

Students who completed the test preparation course performed significantly better in math compared to those who did not.


## Correlation Analysis

Correlation analysis is used to examine the relationship between math, reading, and writing scores.


```python
plt.figure(figsize=(6,4))
sns.heatmap(df[numerical_cols].corr(),annot=True,cmap='coolwarm')
plt.title("Correlation between Scores")
plt.show()
```

## Final Insights and Conclusion

- Reading and writing scores are strongly correlated (0.82).
- Math scores also show strong correlation with reading (0.82) and writing (0.80).
- Students who perform well in one subject tend to perform well in others.
- Test preparation has a positive impact on student performance.
